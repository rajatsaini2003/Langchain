## Langchain
LangChain is a framework for building applications around LLMs.
It gives you abstractions and infrastructure for connecting an LLM to things such as Databases, APIs, Tools,etc

### Agents
An AI agent is an LLM-based system that can decide what actions to take to accomplish a goal.


```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI
def get_weather(city:str) -> str:
    """Get the current weather for a given city."""
    return f"The current weather in {city} is sunny with a temperature of 25°C."
model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
)
agent = create_agent(
    model=model,
    tools=[get_weather],
    system_prompt="You are a helpful assistant that can answer questions and provide information."
)
agent
```

```python
response  = agent.invoke({"messages": [{"role": "user", "content": "What is the weather like in New York?"}]})
```
